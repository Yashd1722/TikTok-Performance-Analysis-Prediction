```python
import pandas as pd
import numpy as np
import re

videos = pd.read_csv("./CSV/creator_videos_cleaned.csv")
print(videos.shape)
```

### Calculate comment-to-like ratio and hook retention rate for each video

the diversity score is calculated based on the action and theme to gather. 
For example ,actions ('vlog','challange') and themes ('tech','Travelling')
each themes and actions are assigned each other multiple times for each creator.


```python
videos['comment_to_like_ratio'] = videos['comment_count'] / videos['like_count'].replace(0, np.nan)
videos['hook_retention_rate'] = videos['like_count'] / videos['view_count'].replace(0, np.nan)
```

```python
pattern = r'(\w+)\s+on\s+(\w+)'

videos['video_description'] = videos['video_description'].str.lower()
videos[['action', 'theme']] = videos['video_description'].str.extract(pattern)
videos['action_theme'] = videos['action'] + '_' + videos['theme']
```

```python
creator_metrics = videos.groupby('creator_id').agg(
    avg_comment_to_like_ratio=('comment_to_like_ratio', 'mean'),
    avg_hook_retention_rate=('hook_retention_rate', 'mean'),
    action_theme_diversity_score=('action_theme', lambda x: x.dropna().nunique())
).reset_index()


videos = videos.merge(creator_metrics, on='creator_id', how='left')
```

```python
videos_subset = videos[['creator_id', 
                        'avg_comment_to_like_ratio', 
                        'avg_hook_retention_rate', 
                        'action_theme_diversity_score']]


videos_subset.to_csv("./CSV/creator_metrics_final.csv", index=False)
print(videos_subset.head())
print(videos_subset.shape)
```
