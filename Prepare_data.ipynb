```python
import pandas as pd
import numpy as np
```

### Load the CSV files

```python
df_profiles = pd.read_csv("./CSV/creator_profiles.csv")
df_videos = pd.read_csv("./CSV/creator_videos.csv")

print("Original creator_profiles.csv shape:", df_profiles.shape)
print("Original creator_profiles.csv columns:", list(df_profiles.columns))
print("\nOriginal creator_videos.csv shape:", df_videos.shape)
print("Original creator_videos.csv columns:", list(df_videos.columns))
print("Original creator_videos.csv columns:", list(df_videos.columns))

```

### Identify missing and Incorrect value from each column

```python
def print_missing_summary(df, df_name):
    missing_counts = df.isnull().sum()
    print(f"\n{df_name} - Columns with missing values:")
    print(missing_counts[missing_counts > 0])

def print_problematic_values(df, df_name, problematic_strings=["?", "0.0.0", "unexpected", ""]):
    print(f"\n{df_name} - Columns with problematic (incorrect) values:")
    for col in df.columns:
        count = df[col].astype(str).isin(problematic_strings).sum()
        if count > 0:
            print(f"  {col}: {count} ")
```

```python
print("===== creator_profiles.csv (before cleaning) =====")
print_missing_summary(df_profiles, "creator_profiles.csv")
print_problematic_values(df_profiles, "creator_profiles.csv")
```

```python
print("\n===== creator_videos.csv (before cleaning) =====")
print_missing_summary(df_videos, "creator_videos.csv")
print_problematic_values(df_videos, "creator_videos.csv")
```

```python
problematic_strings = ["?", "0.0.0", "unexpected", ""]
df_profiles.replace(problematic_strings, np.nan, inplace=True)
df_videos.replace(problematic_strings, np.nan, inplace=True)
```

```python
print("creator_profiles.csv (after replacement, before filling):")

print_missing_summary(df_profiles, "creator_profiles.csv")
print_problematic_values(df_profiles, "creator_profiles.csv")
```

```python
print("\n creator_videos.csv (after replacement, before filling):")
print_missing_summary(df_videos, "creator_videos.csv")
print_problematic_values(df_videos, "creator_videos.csv")
```

Removing or replacing the Replace the string with common value in the columnn and replace the numerical value with mean of that column 

```python
def smart_fillna(df, exclude_cols=[]):
    for col in df.columns:
        if col in exclude_cols:
            continue
        try:
            df[col] = pd.to_numeric(df[col], errors='raise')
            df[col].fillna(df[col].mean(), inplace=True)
        except:
            mode_val = df[col].mode(dropna=True)
            if not mode_val.empty:
                df[col].fillna(mode_val[0], inplace=True)

smart_fillna(df_profiles, exclude_cols=['creator_id'])
smart_fillna(df_videos, exclude_cols=['creator_id', 'video_id'])
```

```python
df_profiles.dropna(subset=['creator_id'], inplace=True)
df_videos.dropna(subset=['creator_id', 'video_id'], inplace=True)
df_profiles.reset_index(drop=True, inplace=True)
df_videos.reset_index(drop=True, inplace=True)
```

```python
df_profiles.dropna(axis=1, how='all', inplace=True)
df_videos.dropna(axis=1, how='all', inplace=True)
```

```python
df_profiles.to_csv("./CSV/creator_profiles_cleaned.csv", index=False)
df_videos.to_csv("./CSV/creator_videos_cleaned.csv", index=False)
```

```python
print("\n creator_profiles_cleaned.csv (after cleaning):")

print("Shape:", df_profiles.shape)
print("Columns:", list(df_profiles.columns))
print_missing_summary(df_profiles, "\n creator_profiles.csv")
print_problematic_values(df_profiles, "\n creator_profiles.csv")
```

```python
print("\n===== creator_videos_cleaned.csv (after cleaning) =====")
print("Shape:", df_videos.shape)
print("Columns:", list(df_videos.columns))
print_missing_summary(df_videos, "creator_videos.csv")
print_problematic_values(df_videos, "creator_videos.csv")
```

```python
print("\nFinal check for missing creator_id/video_id in cleaned files:")
print("creator_profiles_cleaned.csv missing creator_id:", df_profiles['creator_id'].isnull().sum())
print("creator_videos_cleaned.csv missing creator_id:", df_videos['creator_id'].isnull().sum())
print("creator_videos_cleaned.csv missing video_id:", df_videos['video_id'].isnull().sum())
```
